```python
import pandas as pd
df = pd.read_csv("/content/creditcard.csv")
```

```python
df.head()
```

```python
print("Shape :", df.shape)
display(df.info())
display(df.describe())
print(df.isnull().sum().sort_values(ascending = False))
```

```python
class_counts = df["Class"].value_counts()
print(class_counts)
```

```python
print("\nFraud Rate :")
print((class_counts / len(df) * 100))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(data = df, x = "Class")
plt.title("Transactions Class Distribution")
plt.xticks([0,1], ["Legitimate", "Fraud"])
plt.yscale("log")
plt.show()
```

```python
X = df.drop(columns = "Class")
y = df["Class"]
```

```python
print("Features :", X.shape)
print("Target Distribution :\n", y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_full, X_test, y_train_full, y_test = train_test_split(X, y, test_size = 0.20, stratify = y, random_state = 81)
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_train_full, y_train_full, test_size = 0.25, stratify = y_train_full, random_state = 81)
```

```python
print("Train : ", X_train.shape, y_train.value_counts().to_dict())
print("Validation : ", X_val.shape, y_val.value_counts().to_dict())
print("Test : ", X_test.shape, y_test.value_counts().to_dict())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
```

```python
baseline_model = Pipeline(steps =[
    ("imputer", SimpleImputer(strategy = "median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        class_weight = "balanced",
        max_iter = 2000,
        random_state = 81
    ))
])

baseline_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay, roc_auc_score, average_precision_score, precision_recall_curve,RocCurveDisplay,
    PrecisionRecallDisplay
)
```

```python
val_prob = baseline_model.predict_proba(X_val)[:, 1]
val_pred = (val_prob >= 0.5).astype(int)

print(classification_report(y_val, val_pred, digits = 4))
print("ROC-AUC : ", roc_auc_score(y_val, val_prob))
print("PR-AUC : ", average_precision_score(y_val, val_prob))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_val, val_pred, display_labels = ["Legitimate", "Fraud"], cmap = "Blues"
)
plt.title("Validation Confusion Matrix")
plt.show()

PrecisionRecallDisplay.from_predictions(y_val, val_prob)
plt.title("Precision-Recall Curve")
plt.show()
```

```python
precision, recall, thresholds = precision_recall_curve(y_val, val_prob)
threshold_table = pd.DataFrame({
    "threshold" : thresholds,
    "precision" : precision[:-1],
    "recall" : recall[:-1]
})
threshold_table.sort_values(
    by = ["recall", "precision"],
    ascending = False
)
```

```python
target_recall = 0.80
valid_thresholds = threshold_table[
    threshold_table["recall"] >= target_recall
]

chosen_threshold = valid_thresholds.sort_values(
    "precision", ascending = False
).iloc[0]["threshold"]

print("Chosen threshold : ", chosen_threshold)
```

```python
val_pred_tuned = (val_prob >= chosen_threshold).astype(int)

print(classification_report(y_val, val_pred_tuned, digits = 4))
ConfusionMatrixDisplay.from_predictions(
    y_val, val_pred_tuned, display_labels = ["Legitimate", "Fraud"],
    cmap = "Oranges"
)

plt.title(f"Validation Results at Threshold = {chosen_threshold:.3f}")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
```

```python
smote_model = ImbPipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("smote", SMOTE(random_state = 81)),
    ("model", RandomForestClassifier(
        n_estimators = 200,
        max_depth = 12,
        min_samples_leaf = 2,
        n_jobs = -1,
        random_state = 81,
        class_weight = "balanced_subsample"
    ))
])

smote_model.fit(X_train, y_train)
```

```python
smote_val_prob = smote_model.predict_proba(X_val)[:, 1]
smote_val_pred = (smote_val_prob >= 0.5).astype(int)

print(classification_report(y_val, smote_val_pred, digits = 4))
print("ROC-AUC : ", roc_auc_score(y_val, smote_val_prob))
print("PR-AUC", average_precision_score(y_val, smote_val_prob))
```

```python
from sklearn.metrics import (
    average_precision_score, roc_auc_score,
    precision_score, recall_score, f1_score,
    confusion_matrix
)

def evaluate_model(name, model, X_val, y_val, threshold = 0.5):
  probabilities = model.predict_proba(X_val)[:, 1]
  predictions = (probabilities >= threshold).astype(int)
  tn, fp, fn, tp = confusion_matrix(y_val, predictions).ravel()

  return{
      "Model" : name,
      "Threshold" : threshold,
      "PR-AUC" : average_precision_score(y_val, probabilities),
      "ROC-AUC" : roc_auc_score(y_val, probabilities),
      "Recall (fraud caught)" : recall_score(y_val, predictions),
      "Precision (alerts correct)" : precision_score(y_val, predictions, zero_division=0),
      "F1-score" : f1_score(y_val, predictions, zero_division=0),
      "True Positives" : tp,
      "False Negatives (missed fraud)" : fn,
      "False Positives (legitimate flagged)" : fp,
      "True Negatives" : tn
  }

results = pd.DataFrame([
      evaluate_model("Baseline Logistic Regression", baseline_model, X_val, y_val),
      evaluate_model("SMOTE Random Forest", smote_model, X_val, y_val)
  ])

results.sort_values("PR-AUC", ascending = False)
```

```python
final_model = smote_model
final_threshold = chosen_threshold

test_prob = final_model.predict_proba(X_test)[:,1]
test_pred = (test_prob >= final_threshold).astype(int)

print(classification_report(y_test, test_pred, digits = 4))
print("Test ROC-AUC : ", roc_auc_score(y_test, test_prob))
print("Test PR-AUC : ", average_precision_score(y_test, test_prob))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test, test_pred,
    display_labels = ["Legitimate", "Fraud"],
    cmap = "Greens"
)
plt.title("Final Test Confusion Matrix")
plt.show()
```

```python
def predict_fraud(transaction_df, model, threshold):
  """
  transaction_df : DataFrame with exactly the same feature columns as training data.
  """

  probabilities = model.predict_proba(transaction_df)[:,1]
  predictions = (probabilities >= threshold).astype(int)

  results = transaction_df.copy()
  results["fraud_probability"] = probabilities
  results["fraud_predicition"] = predictions
  return results
```
